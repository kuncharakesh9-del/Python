```python
""" String Slicing and Indexing: 
Write a Python program using the above concatenated string as input and performs the 
following tasks:  """

String_1 = "Hello"
String_2 = input("Enter your name")
String_3 = "welcome to Python programming"
String_4 = String_1+" "+ String_2+ ", "+String_3
print(String_4[0])
print(String_4[-1])
print(String_4[5])
print(String_4[-1:11])
print(String_4[::-1])
print(String_4[25:31])

```
